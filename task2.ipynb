```python
from collections import deque

n = int(input("Grid size: "))

start = tuple(map(int, input("Start (row col): ").split()))
goal = tuple(map(int, input("Goal (row col): ").split()))

queue = deque([(start, [start])])
distance = {start: 0}

shortest = []
min_distance = None

while queue:
    pos, path = queue.popleft()


    if pos == goal:
        if min_distance is None:
            min_distance = len(path) - 1

        if len(path) - 1 == min_distance:
            shortest.append(path)

        continue

    r, c = pos

    moves = [
        (r, c + 1),
        (r + 1, c),
        (r, c - 1),
        (r - 1, c)
    ]

    for nr, nc in moves:

        if 0 <= nr < n and 0 <= nc < n:

            new_distance = distance.get((nr, nc), 999)

            if new_distance >= len(path):
                distance[(nr, nc)] = len(path)
                queue.append(((nr, nc), path + [(nr, nc)]))


print("\nAll minimum paths:")

for path in shortest:
    print(" -> ".join(f"{r}{c}" for r, c in path))

print("\nMinimum moves:", min_distance)
print("Number of minimum paths:", len(shortest))
```
