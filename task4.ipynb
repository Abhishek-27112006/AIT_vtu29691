```python
import math

board = [' ' for _ in range(9)]

def print_board():
    for i in range(0, 9, 3):
        print(f" {board[i]} | {board[i+1]} | {board[i+2]} ")
        if i < 6:
            print("---+---+---")
    print()

def evaluate():
    wins = [(0,1,2),(3,4,5),(6,7,8),
            (0,3,6),(1,4,7),(2,5,8),
            (0,4,8),(2,4,6)]

    for a,b,c in wins:
        if board[a] == board[b] == board[c] != ' ':
            return 1 if board[a] == 'O' else -1
    return 0

def minimax(maximize):
    score = evaluate()

    if score != 0 or ' ' not in board:
        return score

    if maximize:
        best = -math.inf
        for i in range(9):
            if board[i] == ' ':
                board[i] = 'O'
                best = max(best, minimax(False))
                board[i] = ' '
        return best
    else:
        best = math.inf
        for i in range(9):
            if board[i] == ' ':
                board[i] = 'X'
                best = min(best, minimax(True))
                board[i] = ' '
        return best

def best_move():
    best = -math.inf
    move = -1

    for i in range(9):
        if board[i] == ' ':
            board[i] = 'O'
            value = minimax(False)
            board[i] = ' '

            if value > best:
                best = value
                move = i
    return move

print("Welcome to Tic-Tac-Toe! You are 'X' and the AI is 'O'.")
print("Positions are indexed from 0 to 8 (top-left to bottom-right).\n")
print_board()

while ' ' in board and evaluate() == 0:
    try:
        move = int(input("Enter your move (0-8): "))

        if move < 0 or move > 8 or board[move] != ' ':
            print("Invalid move! Try again.")
            continue

    except ValueError:
        print("Invalid input! Please enter a number between 0 and 8.")
        continue

    board[move] = 'X'
    print_board()

    if evaluate() == -1 or ' ' not in board:
        break

    print("AI is making a move...")
    board[best_move()] = 'O'
    print_board()

score = evaluate()

if score == 1:
    print("AI wins! Better luck next time.")
elif score == -1:
    print("Congratulations! You won.")
else:
    print("It's a draw!")
```
