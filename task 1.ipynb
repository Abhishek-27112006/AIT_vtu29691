```python
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E', 'G'],
    'G': ['F']
}

visited = set()
path = []

def dfs(node):
    visited.add(node)  # Added: mark node as visited to prevent infinite loops
    path.append(node)
    print("visiting:", node)

    for neighbor in graph[node]:
        if neighbor not in visited:
            dfs(neighbor)

    print("backtracking from:", node)

start = "A"
dfs(start)

print("\nexploration path:", " -> ".join(path))
print("visited chambers:", visited)
```
